```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import imdb
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Embedding,
    SimpleRNN,
    Dense
)

from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
(x_train, y_train), (x_test, y_test) = imdb.load_data()
```

```python
print("Training Shape :", x_train.shape)
print("Testing Shape :", x_test.shape)

print("First Review :", x_train[0])
print("First Label :", y_train[0])
```

```python
print(x_train[0][:15])
```

# Pad Sequences

```python
max_length = 200

x_train = pad_sequences(
    x_train,
    maxlen=max_length
)

x_test = pad_sequences(
    x_test,
    maxlen=max_length
)
```

```python
print(x_train.shape)
```

```python
print(x_test.shape)
```

# Create the Model

```python
vocab_size = 90000
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dropout, Dense

model = Sequential([
    
    # Input Layer
    Input(shape=(max_length,)),

    # Embedding Layer
    Embedding(
        input_dim=vocab_size,
        output_dim=128
    ),

    # Simple RNN Layer
    SimpleRNN(
        units=64,
        activation="tanh"
    ),

    # Dropout Layer
    Dropout(
        rate=0.5
    ),

    # Hidden Dense Layer
    Dense(
        units=32,
        activation="relu"
    ),

    # Output Layer
    Dense(
        units=1,
        activation="sigmoid"
    )

])
```

```python
model.summary()
```

### ============================================
## 7. Compile Model
### ============================================

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

# 8. Train Model

```python
print(x_train.shape)
print(y_train.shape)

print(x_train.dtype)
print(y_train.dtype)

print(x_train.max())
print(x_train.min())
```

```python
print(x_train.max())
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=64,
    validation_split=0.2
)
```

# Evaluate

```python
loss, accuracy = model.evaluate(

    x_test,

    y_test

)

print(accuracy)
```

```python
prediction = model.predict(x_test[:1])
```
