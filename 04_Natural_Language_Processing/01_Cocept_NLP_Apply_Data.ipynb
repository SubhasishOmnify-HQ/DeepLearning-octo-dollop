# Tokenization

```python
text = "Hi Everone! This is Hackers Realm. we are learning Natural Language Processing."
```

```python
text.split(" ")
```

```python
from nltk import sent_tokenize, word_tokenize
```

```python
# split the text into sentences
sent_tokens = sent_tokenize(text)
sent_tokens
```

```python
# split the sentences into word
sent_word = word_tokenize(text)
sent_word
```

# Stemming 

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
word = ('eats')
ps.stem(word)
```

```python
word2 = ('eating')
ps.stem(word2)
```

```python
word3 = ('doing')
ps.stem(word3)
```

```python
text2 = "Hi Everone! This is Hackers Realm. we are learning Natural Language Processing."
```

```python
# split the sentences into word
sent_word2= word_tokenize(text2)
sent_word2
```

```python
stemmed = " ".join(ps.stem(word2) for word2 in sent_word2)
```

```python
stemmed 
```

# Lemmatization

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize('workers')
```

```python
lemmatizer.lemmatize('words')
```

```python
lemmatizer.lemmatize('feet')
```

```python
text3 = "Hi Everone! This is Hackers Realm. we are learning Natural Language Processing. views"
```

```python
sent_word3= word_tokenize(text3)
sent_word3
```

```python
# remove pural forms that conver singual 
lemmatized_sent = " ".join(lemmatizer.lemmatize(word2.lower()) for word2 in sent_word3)
lemmatized_sent
```
