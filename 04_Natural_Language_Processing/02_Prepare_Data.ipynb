# Remove Duplicate Rows

```python
df = pd.read_csv('NLP_Dataset.csv')
```

```python
df.drop_duplicates(inplace=True)
```

# Handle Missing Values

```python
df["text"] = df["text"].fillna("")
```

```python
df
```

# Convert to Lowercase

```python
df["text"] = df["text"].str.lower()
```

```python
df.head()
```

# Remove HTML Tags

```python
def remove_html(text):
    return re.sub(r'<.*?>', '', text)

df["text"] = df["text"].apply(remove_html)
```

```python
df
```

# Remove URLs

```python
def remove_url(text):
    return re.sub(r'https?://\S+|www\.\S+', '', text)

df["text"] = df["text"].apply(remove_url)
```

```python
df
```

# Remove Email Addresses

```python
def remove_email(text):
    return re.sub(r'\S+@\S+', '', text)

df["text"] = df["text"].apply(remove_email)
```

```python
df
```

# Remove @Mentions

```python
def remove_mentions(text):
    return re.sub(r'@\w+', '', text)

df["text"] = df["text"].apply(remove_mentions)
```

```python
df
```

# Remove Emojis

```python
def remove_emoji(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"
        "\U0001F300-\U0001F5FF"
        "\U0001F680-\U0001F6FF"
        "\U0001F1E0-\U0001F1FF"
        "]+",
        flags=re.UNICODE
    )

    return emoji_pattern.sub('', text)

df["text"] = df["text"].apply(remove_emoji)
```

# Remove Punctuation

```python
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

df["text"] = df["text"].apply(remove_punctuation)
```

```python
df
```

# Remove Numbers

```python
def remove_numbers(text):
    return re.sub(r'\d+', '', text)

df["text"] = df["text"].apply(remove_numbers)
```

```python
df
```

# Remove Extra Spaces

```python
def remove_spaces(text):
    return " ".join(text.split())

df["text"] = df["text"].apply(remove_spaces)
```

```python
df
```

# Tokenization

```python
df["tokens"] = df["text"].apply(word_tokenize)
```

```python
df
```

```python
df.tokens
```

# Remove Stopwords

```python
stop_words = set(stopwords.words("english"))
```

```python
stop_words 
```

```python
def remove_stopwords(words):
    return [word for word in words if word not in stop_words]

df["tokens"] = df["tokens"].apply(remove_stopwords)
```

```python
df
```

# Stemming

```python
stemmer = PorterStemmer()

def stem_words(words):
    return [stemmer.stem(word) for word in words]

df["stemmed"] = df["tokens"].apply(stem_words)
```

```python
df
```

# Lemmatization

```python
lemmatizer = WordNetLemmatizer()

def lemma_words(words):
    return [lemmatizer.lemmatize(word) for word in words]

df["lemmatized"] = df["tokens"].apply(lemma_words)
```

```python
df
```

# Join Words Back

```python
df["clean_text"] = df["lemmatized"].apply(lambda x: " ".join(x))
```

```python
df
```

# Remove Frequent Words (Top 5)

```python
from collections import Counter

all_words = " ".join(df["clean_text"]).split()

counter = Counter(all_words)

frequent_words = [word for word, count in counter.most_common(5)]

print(frequent_words)

def remove_frequent(text):
    return " ".join(
        word for word in text.split()
        if word not in frequent_words
    )

df["clean_text"] = df["clean_text"].apply(remove_frequent)
```

```python
df
```

# Remove Rare Words

```python
# split then count each word
word_counts = Counter(all_words)
```

```python
word_counts
```

```python
# Find Rare word
rare_words = [
    word
    for word, count in word_counts.items()
    if count == 1
]
rare_words
```

```python
# Remove Rare Words
def remove_rare_words(text):
    return " ".join(
        word
        for word in text.split()
        if word not in rare_words
    )

df["clean_text"] = df["text"].apply(remove_rare_words)
```

```python
df
```
