# 1. Bag of Word(BoW)

```python
from sklearn.feature_extraction.text import CountVectorizer

text = [
    "I love NLP",
    "I love Python"
]

cv = CountVectorizer()

X = cv.fit_transform(text)

print(cv.get_feature_names_out())
print(X.toarray())
```

# 2. CountVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer

text = [
    "Machine learning is amazing",
    "Learning NLP is fun"
]

cv = CountVectorizer(stop_words='english')

X = cv.fit_transform(text)

print(cv.get_feature_names_out())
print(X.toarray())
```

# 3. N-Grams

```python
cv = CountVectorizer(ngram_range=(2,2))

X = cv.fit_transform(["I love NLP"])

print(cv.get_feature_names_out())
```

# 4. TF-IDF (Term Frequency–Inverse Document Frequency)

```python
from sklearn.feature_extraction.text import TfidfVectorizer

text = [
    "I love NLP",
    "I love Python"
]

tfidf = TfidfVectorizer()

X = tfidf.fit_transform(text)

print(tfidf.get_feature_names_out())
print(X.toarray())
```

# 5. Word2Vec

### pip install gensim

```python
from gensim.models import Word2Vec

sentences = [
    ["i", "love", "nlp"],
    ["nlp", "is", "interesting"],
    ["i", "love", "machine", "learning"],
    ["deep", "learning", "uses", "neural", "networks"]
]

model = Word2Vec(
    sentences,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4
)
```

```python
# Get a Word Vector
vector = model.wv["nlp"]
print(vector)
```

```python
# Find Similar Words
model.wv.most_similar("nlp")
```

```python
# Vocabulary
print(model.wv.index_to_key)
```

```python
# Word Similarity
print(model.wv.similarity("nlp", "learning"))
```

```python
# Save the Model
model.save("word2vec.model")
```

# 6. FastText

```python
import pandas as pd
from gensim.models import FastText
```

```python
data = {
    "review": [
        "I love NLP",
        "Machine learning is amazing",
        "Deep learning uses neural networks",
        "Python is great for AI",
        "Artificial Intelligence is the future"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Train Model
model = FastText(
    sentences=data,
    vector_size=100,
    window=5,
    min_count=1,
    workers=4,
    epochs=20
)
```
